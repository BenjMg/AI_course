# TP 5 - Développement d'un assistant LLM en python "light"

## Partie 1 - Création d'un client LLM

L'objectif est de créer une instance light d'un assitant LLM à partir du SDK OpenAI

#### Q1. 
a. Créer une class **MyLLM** qui prend en paramètre un system_prompt et un modèle, et instancie le client.

b. Définir une fonction **chat** qui prend un text utilisateur et renvoie une réponse. 

c. On pourra utiliser la méthode vu en TD ou bien en utilisant la méthode **completions** : https://vercel.com/docs/ai-gateway/sdks-and-apis/openai-chat-completions/streaming

#### Q2. 
Tester différents **system_prompt** et observer leur influence sur les réponses du modèle. Le comportement de l'assistant change-t-il selon les instructions fournies ?

## Partie 2 — Gestion de l'historique de la conversation

#### Q3. 
Modifier la classe MyLLM pour conserver l'historique des messages échangés entre l'utilisateur et l'assistant. On pourra utiliser le rôle **assistant** pour désigner le modèle. 

#### Q4.
Modifier la méthode chat() afin que chaque nouvelle requête transmette au modèle l'historique de la conversation, et pas uniquement le dernier message.

#### Q5.
Ajouter une méthode new_chat() permettant de réinitialiser l'historique de la conversation sans modifier le modèle ni le system_prompt.


#### Q6.
Que se passe-t-il quand la taille de l'historique augmente par rapport au contexte du modèle ? Proposer une solution pour gérer le contexte par rapport à l'historique de la conversation.


## Partie 3 - Gestion de plusieurs conversation 

#### Q7.
Proposer une solution pour la gestion de plusieurs conversation. Modifier la fonction **new_chat** pour générer une nouvelle conversation sans perdre l'historique de la conversation précédente. Créer une fonction **delete** pour supprimer une conversation. Et créer une fonction **change_chat** pour basculer d'une conversaiton à une autre.
