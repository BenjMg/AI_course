# TP 6 - Intégration d'un RAG pour améliorer notre client LLM

#### Q1. Création d'une class **MyRAG**

a. Créer une class **MyRAG** qui prend en input un modèle d'embedding et instancie un client Qdrant

b. Définir une fonction **create_collection** qui permet de créer une nouvelle collection. Pensez à verifier que la collection existe déjà. 

c. Définir une fonction **add_points** qui prend en input un text et ajoute sa représentation (embedding + payload) à une collection. On pourra gérer les id Qdrant à l'aide d'un uuid. Comment gérer les doublons dans ce cas là ? 

d. Enfin définir une fonction retrieval qui prend en input une querry, un collection_name et un top_k et retourne les instances les plus proches. 

/!\ On part du principe que chaque nouvelle collection utilise le même modèle embedding. On pourra aussi garder les documents originaux, soit directement dans la payload, soit dans une base à part. 

#### Q2. Modification de l'agent **MyLLM**

a. A partir de l'agent **MyLLM** définit la semaine denière, ajouter une option dans l'initialisation du modèle et/ou dans la fonction chat pour que le l'agent interoge le RAG et une collection spécifique quand l'utilisateur pose une question. 

#### Q3. Spécialiser votre agent **MyLLM** avec **MyRAG**

a. Choisir un sujet et instancier votre agent pour qu'il se spécialise sur ce sujet (system prompt). 

b. Collecter des documents spécifique à ce sujet et reproduire le process jusqu'à l'ingestion des documents dans une colleciton spécifique. 

c. Comparer l'efficacité de votre agent avec et sans l'utilisation du RAG. Vous pouvez aussi contraindre l'agent à ne répondre que en se bassant sur son contexte (system prompt et/ou user prompt)

/!\ Attention à la gestion du contexte avec les documents issus du RAG et leur longueur.
